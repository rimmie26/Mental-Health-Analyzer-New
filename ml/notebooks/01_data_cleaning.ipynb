Handling Missing Values and Duplicate rows

```python
missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Percentage": (df.isnull().sum()/len(df))*100
})

missing = missing[missing["Missing Count"]>0]

missing.sort_values("Percentage", ascending=False)
```

Removal of UNnecessary Columns 

```python
clean_df = df.copy()

columns_to_remove = [
    "id",
    "City",
    "Profession",
    "Work Pressure",
    "Job Satisfaction"
]

clean_df.drop(columns=columns_to_remove, inplace=True)
```

```python
clean_df
```

```python
clean_df[clean_df["Financial Stress"] == "?"]
```

```python
clean_df["Financial Stress"] = clean_df["Financial Stress"].replace("?", np.nan)

clean_df = clean_df.dropna(subset=["Financial Stress"])

clean_df["Financial Stress"] = clean_df["Financial Stress"].astype(int)
```

```python
clean_df[clean_df["Academic Pressure"] == 0]
```

```python
print("Rows with Academic Pressure = 0:",
      (clean_df["Academic Pressure"] == 0).sum())
```

```python
clean_df = clean_df[clean_df["Academic Pressure"] != 0]
```

```python
clean_df["Academic Pressure"].value_counts().sort_index()
```

### Cleaning Step – Academic Pressure

The survey uses a Likert scale from **1 to 5** for Academic Pressure.

Nine records contained a value of **0**, which falls outside the valid range and is considered an invalid response.

These rows were removed from the dataset to maintain data quality.

```python
clean_df[clean_df["Sleep Duration"] == "Others"]
```

```python
clean_df = clean_df[clean_df["Sleep Duration"] != "Others"]
```

```python
clean_df[clean_df["Dietary Habits"] == "Others"]
```

```python
clean_df = clean_df[clean_df["Dietary Habits"] != "Others"]
```

```python
clean_df[clean_df["Age"] > 35]
```

```python
clean_df = clean_df[clean_df["Age"] < 35]
```

```python
clean_df["Depression"].value_counts()
```

```python
output_path = r"D:\Project\Mental-Health-Analyzer-New\ml\data\processed\cleaned_kaggle.csv"

clean_df.to_csv(output_path, index=False)

print("✅ Dataset saved successfully!")
```

## Summary of Data Cleaning

### Key Cleaning Operations
- Removed non-informative columns (`id`, `City`, `Profession`, `Work Pressure`, `Job Satisfaction`).
- Removed invalid responses in `Financial Stress`.
- Removed invalid values in `Academic Pressure` and `Study Satisfaction`.
- Removed ambiguous `Others` responses from `Sleep Duration` and `Dietary Habits`.
- Removed inconsistent age records.

### Outcome
The cleaned dataset is ready for EDA, feature engineering, predictive modeling, and the wellness score framework.
