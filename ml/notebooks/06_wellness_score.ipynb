## Objective

The objective of this notebook is to convert raw student responses into meaningful wellness scores.

Instead of presenting only a binary depression prediction, the system evaluates different dimensions of student well-being.

The wellness score provides an interpretable assessment across multiple categories such as academics, sleep, lifestyle, mental health, and financial well-being.

These scores will later be used to generate personalized recommendations.

```python
HEALTH_INDICATORS = {
    "Academic Health": [
        "Academic Pressure",
        "Study Satisfaction",
        "CGPA",
        "Work/Study Hours"
    ],

    "Sleep Health": [
        "Sleep Duration"
    ],

    "Lifestyle Health": [
        "Dietary Habits"
    ],

    "Mental Well-being": [
        "Have you ever had suicidal thoughts ?"
    ],

    "Financial Well-being": [
        "Financial Stress"
    ]
}
```

```python
academic_pressure_score = {

    1:100,
    2:80,
    3:60,
    4:40,
    5:20

}
```

```python
study_satisfaction_score = {

    1:20,
    2:40,
    3:60,
    4:80,
    5:100

}
```

```python
sleep_score = {

    0:25,   # <5 hours
    1:60,   # 5-6 hours
    2:100,  # 7-8 hours
    3:85    # >8 hours

}
```

```python
diet_score = {

    0:30,
    1:70,
    2:100

}
```

```python
financial_score = {

    1:100,
    2:80,
    3:60,
    4:40,
    5:20

}
```

```python
mental_score = {

    0:100,
    1:20

}
```

```python
def cgpa_score(cgpa):
    return (cgpa / 10) * 100
```

```python
def study_hours_score(hours):

    if hours <= 2:
        return 60

    elif hours <= 5:
        return 100

    elif hours <= 8:
        return 80

    else:
        return 50
```

```python
def calculate_wellness(student):
    
    academic = np.mean([
        academic_pressure_score[student["Academic Pressure"]],
        study_satisfaction_score[student["Study Satisfaction"]],
        cgpa_score(student["CGPA"]),
        study_hours_score(student["Work/Study Hours"])
    ])

    sleep = sleep_score[student["Sleep Duration"]]

    lifestyle = diet_score[student["Dietary Habits"]]

    mental = mental_score[
        student["Have you ever had suicidal thoughts ?"]
    ]

    financial = financial_score[
        student["Financial Stress"]
    ]

    return {

        "Academic": academic,
        "Sleep": sleep,
        "Lifestyle": lifestyle,
        "Mental": mental,
        "Financial": financial

    }
```
